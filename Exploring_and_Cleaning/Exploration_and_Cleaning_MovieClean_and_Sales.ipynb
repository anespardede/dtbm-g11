<h3>Title-year matching experiments</h3>

```python
# Prepare title and year columns for matching

sales_clean["title_match"] = (
    sales_clean["title"]
    .astype("string")
    .str.strip()
    .str.lower()
)

metaclean_clean["title_match"] = (
    metaclean_clean["title"]
    .astype("string")
    .str.strip()
    .str.lower()
)

metaclean_clean["year_match"] = metaclean_clean["RelDate"].dt.year

print("Sales title_match created:", "title_match" in sales_clean.columns)
print("MetaClean title_match created:", "title_match" in metaclean_clean.columns)
print("MetaClean year_match created:", "year_match" in metaclean_clean.columns)
```

```python
#Count unique Sales-MetaClean title/year matches

sales_match_test = sales_clean.merge(
    metaclean_clean[["title_match", "year_match", "url"]],
    left_on=["title_match", "year"],
    right_on=["title_match", "year_match"],
    how="inner",
    suffixes=("_sales", "_meta")
)

print("Total matched rows:", len(sales_match_test))

print(
    "Unique Sales URLs matched:",
    sales_match_test["url_sales"].nunique()
)

print(
    "Unique MetaClean URLs matched:",
    sales_match_test["url_meta"].nunique()
)
```

```python
# Find title + year combinations that create multiple matches

# Test title + year matching

match_counts = (
    sales_match_test
    .groupby(["title_match", "year"])
    .agg(
        sales_urls=("url_sales", "nunique"),
        metacritic_urls=("url_meta", "nunique")
    )
    .reset_index()
)

problem_matches = match_counts[
    (match_counts["sales_urls"] > 1) |
    (match_counts["metacritic_urls"] > 1)
]

print("Problematic title + year matches:", len(problem_matches))

print(problem_matches.to_string(index=False))
```
